### Data Cleaning (Replace this with more specific notes!)

```python
#Clean Price
#Convert to string first, then replace
df['price'] = df['price'].astype(str).str.replace('₹', '').str.replace(',', '')
```

```python
# Convert to numeric, and if there's any weird text, turn it into NaN instead of crashing
df['price'] = pd.to_numeric(df['price'], errors='coerce')
```

```python
#Handle missing ratings with the median (median is used cuz of outlayers (less affected than mean) & ratings is skewed distribution )
df['rating'] = df['rating'].fillna(df['rating'].median())
```

```python
#Remove any rows that ended up with NaN in our main columns
df.dropna(subset=['price', 'ram', 'battery'], inplace=True)
```

```python
import re

def extract_storage(text):
    if pd.isna(text): return np.nan
    match = re.search(r'(\d+)\s*GB inbuilt', str(text))
    if match: return int(match.group(1))
    match = re.search(r'(\d+)\s*TB inbuilt', str(text))
    if match: return int(match.group(1)) * 1024
    return np.nan

def extract_ram(text):
    if pd.isna(text): return np.nan
    match = re.search(r'(\d+)\s*GB RAM', str(text))
    if match: return int(match.group(1))
    return np.nan

def extract_processor_brand(text):
    if pd.isna(text): return "Unknown"
    text = str(text).lower()
    if 'snapdragon' in text: return 'Snapdragon'
    if 'dimensity' in text: return 'Dimensity'
    if 'bionic' in text: return 'Apple Bionic'
    if 'helio' in text: return 'Helio'
    if 'exynos' in text: return 'Exynos'
    if 'unisoc' in text: return 'Unisoc'
    if 'tensor' in text: return 'Google Tensor'
    return 'Other'

# remove unrealistic prices
df = df[df['price'] > 2000]

df['brand'] = df['model'].str.split().str[0]
df['ram_gb'] = df['ram'].apply(extract_ram)
df['storage_gb'] = df['ram'].apply(extract_storage)
df['ram_per_1k_price'] = (df['ram_gb'] / df['price']) * 1000
df['storage_per_1k_price'] = (df['storage_gb'] / df['price']) * 1000
df['processor_brand'] = df['processor'].apply(extract_processor_brand)
```

```python
# Final Check

print(df[['model', 'price', 'ram', 'battery']].head())
```

### Which brands offer the best specifications for price?

```python
df['brand'] = df['model'].str.split().str[0] #split cuts the model to words (Samsung Galaxy S23) , str[0] takes the first word

brand_stats = df.groupby('brand').agg({ 
    'ram_per_1k_price': 'mean',
    'storage_per_1k_price': 'mean',
    'model': 'count'
}).rename(columns={'model': 'count'})  #aggregation , counts how many models in this brand

top_value_brands = brand_stats[brand_stats['count'] > 5].sort_values(by='ram_per_1k_price', ascending=False) #delete brands that has less than 5 models , then arrange them in as descending order according to the ram_per_1k_price

print(top_value_brands[['ram_per_1k_price', 'storage_per_1k_price']].head(10))
```

### Compare budget vs flagship devices

```python
df['segment'] = 'Mid-Range' #initially supposed that all mobiles are mid-range
df.loc[df['price'] < 15000, 'segment'] = 'Budget' #locate mobiles with price less than 15000 in Budget
df.loc[df['price'] > 50000, 'segment'] = 'Flagship' #locate mobiles with price more than 50000 in Flagship

compare_df = df[df['segment'] != 'Mid-Range'] #compare lowest and highest neglecting midrange

fig, ax = plt.subplots(1, 2, figsize=(15, 6)) #draws 2 graphs on 1 line , ax matrix has 2 ax[0] & ax[1]

sns.barplot(x='segment', y='ram_gb', data=compare_df, ax=ax[0], palette='coolwarm', ci=None) # ci=none deletes small black lines that describes the errors 
ax[0].set_title('Average RAM: Budget vs Flagship')
ax[0].set_ylabel('RAM (GB)')

sns.barplot(x='segment', y='rating', data=compare_df, ax=ax[1], palette='viridis', ci=None)
ax[1].set_title('Average Rating: Budget vs Flagship')
ax[1].set_ylabel('Rating (out of 100)')

plt.tight_layout()
plt.show()
```

### Identify best value-for-money products

```python

df['value_score'] = (df['ram_gb'] * df['rating']) / df['price']

top_vfm = df.sort_values(by='value_score', ascending=False).head(5)

plt.figure(figsize=(10,6))
sns.barplot(x='value_score', y='brand', data=df.groupby('brand')['value_score'].mean().sort_values(ascending=False).reset_index().head(10), palette='Greens_r')
plt.title('Top 10 Brands by Value-for-Money Score')
plt.xlabel('Average Value Score (Higher = Better Value)')
plt.show()

print("The Top 5 Value-for-Money Smartphones are:")
print(top_vfm[['model', 'price', 'ram_gb', 'rating']])
```

### III. Important Considerations & Limitations

It's crucial to interpret these findings with a few caveats in mind:

1.  **Data Scope:** This analysis is based on a specific dataset, likely from a particular retailer or region. Therefore, these conclusions might not be universally applicable to the global smartphone market or all regional nuances.

2.  **Missing Details:** Our analysis couldn't account for other important features that influence phone prices and user perception, such as camera megapixels, display refresh rates, water resistance, or software support duration.

3.  **Correlation vs. Causation:** Remember, strong correlations don't necessarily mean one thing *causes* another. For example, while high RAM and high prices go hand-in-hand, it doesn't mean simply increasing an old phone's RAM will make it more valuable. High-end phones simply tend to bundle many premium features, including advanced hardware.
